```python
ruta_estandar= []
ruta_urgente= []
registro_paquetes= {}
menu_activo=True 

while menu_activo:
    
    print("SISTEMA AVANZADO FASTBOX")

    print("==================== Menú Principal ===================")

    print("1. Registrar nuevo paquete")
    print("2. Despachar paquete")
    print("3. Ver estado detallado de las rutas")
    print("4. Salir")

    seguir_pidiendo= True

    while seguir_pidiendo:
        opcion_texto = input("¿Cuál es la opción del menú que deseas comprobar (1,2,3 o 4): ")
        if opcion_texto.isnumeric():
            opcion=int(opcion_texto)
            if opcion >= 1 and opcion <= 4:
                seguir_pidiendo= False
            else:
                print("Tu número no está entre las opciones disponibles\n Intentalo de nuevo, por favor")
        else:
            print("Respuesta no valida\nPor favor, escribe un número")
    if opcion == 1:
        
        print("Has escogido la opción de registrar un nuevo paquete")
        
        identificacion_paquete= input("Añade el ID del paquete: ")
        while identificacion_paquete in registro_paquetes:
            print("El ID que has añadido ya está registrado a otro paquete")
            identificacion_paquete= input("Añade el ID del paquete: ")

        seguir_pidiendo_peso= True
        while seguir_pidiendo_peso:
            peso_texto= input("¿Cuál es el peso del paquete en kg: ")
            if peso_texto.isnumeric():
                peso=int(peso_texto)
                if peso> 0:
                    print ("El paquete con ID", identificacion_paquete, "pesa", peso, "kg" )
                    seguir_pidiendo_peso= False
                else:
                    print("El peso que has puesto no es valido.\nIntentalo de nuevo.\nAñade el peso del paquete:")
            else:
                print("El peso del paquete debe ser un número entero")
        
        seguir_pidiendo_envio= True
        while seguir_pidiendo_envio:
            tipo_envio_texto= input("¿Qué tipo de envio vas a querer? La opción 1(Estándar) o la opción 2(Urgente)")
            if tipo_envio_texto.isnumeric():
                tipo_envio= int(tipo_envio_texto)
                if tipo_envio==1 or tipo_envio==2:
                    if tipo_envio==1:
                        ruta_estandar.append(identificacion_paquete)
                    else:
                        ruta_urgente.append(identificacion_paquete)
                    registro_paquetes[identificacion_paquete]=peso
                    print("Contenido actual del registro:", registro_paquetes)
                    seguir_pidiendo_envio= False 
                else:
                    print("Respuesta no valida\nPor favor, selecione la opción 1 de estándar o la 2 de urgente")
            else:
                print("Por favor, escribe un número valido")
        
        
    if opcion == 2:
        
        print ("Has escogido la opción de despachar un paquete")
        
        despachar= True 
        while despachar:
            ruta_despachar_texto= input("¿Qué ruta deseas despachar la opción 1 (Estándar) o la opción 2 (Urgente)?")
            if ruta_despachar_texto.isnumeric():
                ruta_despachar= int(ruta_despachar_texto)
                if ruta_despachar==1 or ruta_despachar==2:
                    despachar=False
                else:
                    print ("Respuesta no valida.\nPor favor, seleciona la opción 1 (Estándar) o la opción 2 (Urgente)")
            else:
                print ("Por favor escribe un número entero")

        if ruta_despachar==1:
            if len(ruta_estandar) == 0:
                print("La ruta estándar está vacía, no tiene ningún paquete guardado")
            else:
                identificacion_paquete = ruta_estandar.pop(0)
               
                peso = registro_paquetes[identificacion_paquete]
                print("Paquete despachado: ", identificacion_paquete, "\nPeso: ", peso, "kg")
                del registro_paquetes[identificacion_paquete]
        elif ruta_despachar==2:
            if len(ruta_urgente) == 0:
                print("La ruta urgente está vacía, no tiene ningún paquete guardado")
            else:
                identificacion_paquete =ruta_urgente.pop()
                peso = registro_paquetes[identificacion_paquete]
                print("Paquete despachado: ", identificacion_paquete, "\nPeso: ", peso, "kg")
                del registro_paquetes[identificacion_paquete]
        
    if opcion == 3:
        
        print("Has escogido la opción de ver el estado de las rutas de los paquetes")

        print("\n Los paquetes de la ruta estándar: ")
        if len(ruta_estandar)== 0:
            print("La ruta estándar se encuentra vacía")
        else:
            for identificacion_paquete in ruta_estandar:
                peso= registro_paquetes[identificacion_paquete]
                print("ID: ", identificacion_paquete, "| Peso: ", peso, "kg")

        print("\n Los paquetes de la ruta urgente: ")
        if len(ruta_urgente)==0:
            print("La ruta urgente se encuentra vacía")
        else:
            for posicion in range (len(ruta_urgente)-1, -1, -1):
                identificacion_paquete= ruta_urgente[posicion]
                peso= registro_paquetes[identificacion_paquete]
                if posicion == len(ruta_urgente)-1:
                    print("ID: ", identificacion_paquete, "| Peso: ", peso, "kg (Cima)")
                else:
                    print("ID: ", identificacion_paquete, "| Peso: ", peso, "kg")

                    

        
    if opcion == 4:
        print("Gracias por usar FASTBOX.\n ¡Hasta pronto!\nQue tenga un buen día")
        menu_activo=False
```
